```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"D:\development\datasets\titanic\titanicData\tested.csv")
```

```python
df.head()
```

```python
print(pd.crosstab(df["Survived"], df["Sex"]))
```

```python
df40 = df[df["Age"]>40]
count = df40["Survived"].value_counts()
```

```python
plt.bar(count.index, count.values, color = ["red", "green"])
plt.xticks([0, 1], ["Didnt Survive", "Survived"])
plt.ylabel("Passengers")
plt.title("Survivors older than 40")
plt.show
        
```

```python
survivors = df[df["Survived"]==1]
count = survivors["Pclass"].value_counts().sort_index()
```

```python
plt.pie(count.values, labels=["1 Class", "2 Class", "3 Class"],
        autopct="%1.1f%%", startangle=90, colors=["gold", "silver", "brown"], wedgeprops=dict(width=0.4))
plt.title("Survivors Percentage By Class")
plt.show()     
```

```python
ss = df[df["Survived"]==1]
count_sex = ss["Sex"].value_counts()
```

```python
plt.pie(count_sex.values, labels=count_sex.index, autopct="%1.1f%%",
        colors=["pink", "lightblue"], startangle=90)
plt.title("Survivors By Sex")
plt.show()
```
