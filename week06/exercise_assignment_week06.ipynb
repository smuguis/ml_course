```python
df['features'] = (
    df['listed_in'].fillna('') + ' ' +
    df['description'].fillna('')
)
```

```python
df.head()
```

```python
base = df.loc[df['title'].str.lower() == 'squid game'.lower(), 'features'].values[0]
print(base)
```

```python
import re

def words(text):
    text = text.lower()
    text = re.sub(r'[^a-z0-9\s,]', '', text)
    return set(text.split())

def jaccard_similarity(a, b):
    if not a or not b:
        return 0
    inter = len(a & b)
    union = len(a | b)
    return inter / union
```

```python
words_base = words(base)

df['similarity'] = df['features'].apply(lambda x: jaccard_similarity(words_base, words(x)))
```

```python
recs = (
    df[df['title'].str.lower() != 'squid game'] 
    .sort_values('similarity', ascending=False)
    .head(8)[['title', 'type', 'listed_in', 'description', 'similarity']]
)

print(recs)

```
